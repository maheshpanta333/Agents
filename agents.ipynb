```python
import math
from langchain_core.tools import tool
```

```python
@tool
def calculate_area_of_circle(radius:float)->float:
    """Use this agent wheneever user asks to calculate the area of the circle when the radius is given in float value then return the area of the circle in float.
    Calculate the area of the circle given its raidus!"""
    return math.pi*(radius**2)

tools=[calculate_area_of_circle]
```

```python
from langchain.agents import create_agent
```

```python
agent_executor=create_agent(model=llm,tools=tools)
user_query="if i have a  donut circular in shappe with 3m raidus what would be its surface area?"
print("....Starting the agent....")
response=agent_executor.invoke({
    "messages":[("user",user_query)]
})
#this would be the response
print("=============FINAL RESPONSE============")
print(response["messages"][-1].content)
```
